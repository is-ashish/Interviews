------------------- parallel-------------------------------

```python
import threading
import time

def download(url):
    print(f"start download: {url}")
    time.sleep(2)
    print(f"Finished download: {url}")

urls = ["url-1", "url-2", "url-3"]

threads= []

for url in urls:
    thread =  threading.Thread(target=download, args=(url,))
    threads.append(thread)
    thread.start()
    
# wait for all threads to finish
for thread in threads:
    thread.join()

print("All Download Completed")
```

--------------------------- Sequencial -------------------------------

```python
import time

def download(url):
    print(f"start download: {url}")
    time.sleep(5)
    print(f"Finished download: {url}")

urls = ["url-1", "url-2", "url-3"]

for url in urls:
    print(f"----------- {url}")
    download(url)
    print(f"----------- {url}")

print("All Download Completed")
```

----------------- Concurrent.futures.ThreadPoolExecutor, which is a cleaner modern way to handle threads in Python ---------------------------------

```python
import time
from concurrent.futures import ThreadPoolExecutor

def download(url):
    print(f"start download: {url} \n")
    time.sleep(5)
    print(f"Finished download: {url} \n")

urls = ["url-1", "url-2", "url-3"]

start = time.time()

# Use ThreadPoolExecutor to manage threads
with ThreadPoolExecutor(max_workers=3) as executor:
    executor.map(download, urls)

end = time.time()
print("All Download Completed")
print(f"Total time taken: {end - start:.2f} seconds")
```

```python
import asyncio

async def download(url):
    print(f"start download: {url}")
    await asyncio.sleep(5)
    print(f"Finished download: {url}")

async def main():
    urls = ["url-1", "url-2", "url-3"]
    await asyncio.gather(*(download(url) for url in urls))
    print("All Download Completed")

# For scripts: use asyncio.run(main())
# For Jupyter/IPython: just write `await main()`
await main()
```
