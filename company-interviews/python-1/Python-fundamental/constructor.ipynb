```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def greet(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."
    
p1 = Person("ashish", 30)
p2 = Person("Siddhi", 4)

# print(p1.name, p1.age)
# print(p2.name, p2.age)
print(p1.greet())
print(p2.greet())        
```
