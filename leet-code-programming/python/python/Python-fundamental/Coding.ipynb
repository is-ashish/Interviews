```python
def names_decorator(function):
    def wrapper(*args, **kwargs):
        # Capitalize all positional arguments (strings)
        args = [arg.capitalize() for arg in args]
        
        # Capitalize all keyword argument values (strings)
        kwargs = {k: v.capitalize() for k, v in kwargs.items()}
        
        # Pass the modified arguments into the function
        return function(*args, **kwargs)
    return wrapper

@names_decorator
def say_hello(name1, name2):
    return f'Hello {name1}! Hello {name2}!'

print(say_hello('sara', 'ansh')) 
# Output => 'Hello Sara! Hello Ansh!'
```

Top K frequent elements -: 

Given an integer array nums and an integer k, return the k most frequent elements. You may return the answer in any order.

```python
nums = [1,2,1,2,1,2,3,1,3,2, 2]
k =2
def topFrequency(nums, k):
    freq = {}
    
    for num in nums:
        if num in freq:
            freq[num] +=1
        else:
            freq[num] =1
    
    # lets short by keys
    sorted_key =  sorted(freq.items(), key=lambda x: x[1], reverse=True)
    return [item[0] for item in sorted_key[:k]]

print(topFrequency(nums, k))
```

2. Hash Map + Heap (Optimal)
	◦ Count frequencies.
	◦ Use a min-heap of size k to keep track of top elements.
	◦ Time: O(n log k).
3. Bucket Sort (Most Optimal)
	◦ Count frequencies.
	◦ Place numbers into buckets indexed by frequency.
	◦ Traverse buckets from high to low until you collect k elements.
	◦ Time: O(n).

```python
import heapq
nums = [1,2,1,2,1,2,3,1,3,2, 2]
k =2
def minHeapTopK(nums, k):
    freq = {}
    
    for num in nums:
        freq[num] = freq.get(num,0)+1
    # Step 2: Use a min-heap of size k
    heap = []
    for num, count in freq.items():
        heapq.heappush(heap, (count, num))
        if len(heap)>k:
            heapq.heappop(heap)
    return [num for _, num in heap]
        

print(minHeapTopK(nums, k))
```

```python
#Buckt short 
def topKwithBucketshort(nums, k):
    freq = {}
    for num in nums:
        freq[num] = freq.get(num,0)+1
    
    # Step 2: Create buckets (size = len(nums) + 1)
    buckets = [[] for _ in range(len(nums) + 1)]
    for num, count in freq.items():
        buckets[count].append(num)
    print(buckets)

    # # Step 3: Traverse buckets from high to low
    res = []
    for i in range(len(buckets) - 1, 0, -1):  # start from highest frequency
        for num in buckets[i]:
            res.append(num)
            if len(res) == k:
                return res
    return res

nums = [1,2,1,2,1,2,3,1,3,2,2]
k =2

print(topKwithBucketshort(nums, k))
```

3. Maximum Average Subarray I (LeetCode 643)
The Problem: Given an arrsay nums consisting of 
n
n elements, and an integer k, find a contiguous subarray whose length is equal to k that has the maximum average value and return this value.
Example: nums = [1, 12, -5, -6, 50, 3], k = 4 
→
→ Result: 12.75 (Subarray is [12, -5, -6, 50])
Interview Insight: Your introduction to the Fixed-Size Sliding Window.

```python
import math
def maxContiguousAverage(nums, k):
    avgMax = -math.inf
    currSum = 0
    for i in range(len(nums)):
        if i < k:
            currSum = currSum+nums[i]
        else:
            break
    avg = currSum/k
    avgMax = max(avgMax, avg)
    
    for j in range(1,len(nums)-k+1):
        left = nums[j-1]
        right = nums[k+j-1]
        currSum = currSum-left+right
        avg = currSum/k
        avgMax = max(avgMax, avg)
    return avgMax
    


nums = [4,2,1,3,3]
k = 2 # expected = 3
print(maxContiguousAverage(nums, k)) 
```

5. Minimum Window Substring (LeetCode 76)
The Problem: Given two strings s and t, return the minimum window substring of s such that every character in t (including duplicates) is included in the window.
Example: s = "ADOBECODEBANC", t = "ABC" 
→
→ Result: "BANC"
Interview Insight: This is a Hard-level question. It forces you to combine sliding window with the frequency counting logic you just mastered.

```python
def checkCharacterSubset(s, t):
    freq_t = {}
    freq_s = {}
    for i in range(len(t)):
        freq_t[t[i]] = freq_t.get(t[i],0)+1
    for j in range(len(s)):
        freq_s[s[j]] = freq_s.get(s[j], 0)+1

    # Compare counts
    for k in freq_t.keys():
        if k not in freq_s or freq_s[k]<freq_t[k]:
            return False
    return True

def minimumWindowSubstring(s, t):
    
    left=0
    min_len =math.inf
    result = ""
    for right in range(len(s)):
        # expand window
        while(left<=right and checkCharacterSubset(s[left:right+1], t)): #once reach = ADOBEC
            #update minimun
            if right-left+1<min_len:
                min_len = right-left+1
                result = s[left:right+1]
            left +=1
    return min_len
        
s,t = "ADOBECODEBANC", "ABC" 
print(minimumWindowSubstring(s,t))
```

Given two strings s1 and s2, return true if s2 contains a permutation of s1, or false otherwise.

In other words, return true if one of s1's permutations is the substring of s2.

```python
There is an integer array nums sorted in ascending order (with distinct values).

Prior to being passed to your function, nums is possibly left rotated at an unknown index k (1 <= k < nums.length) such that the resulting array is [nums[k], nums[k+1], ..., nums[n-1], nums[0], nums[1], ..., nums[k-1]] (0-indexed). For example, [0,1,2,4,5,6,7] might be left rotated by 3 indices and become [4,5,6,7,0,1,2].

Given the array nums after the possible rotation and an integer target, return the index of target if it is in nums, or -1 if it is not in nums.

You must write an algorithm with O(log n) runtime complexity.
```

```python
def search(nums, target):
    left, right = 0, len(nums) - 1
    
    while left <= right:
        mid = (left + right) // 2
        
        if nums[mid] == target:
            return mid
        
        # Check if left half is sorted
        if nums[left] <= nums[mid]:
            if nums[left] <= target < nums[mid]:
                right = mid - 1
            else:
                left = mid + 1
        else:
            # Right half is sorted
            if nums[mid] < target <= nums[right]:
                left = mid + 1
            else:
                right = mid - 1
    
    return -1


# Example
nums = [4,5,6,7,0,1,2]
target = 0
print(search(nums, target))  # Output: 4
```

Suppose an array of length n sorted in ascending order is rotated between 1 and n times. For example, the array nums = [0,1,2,4,5,6,7] might become:

[4,5,6,7,0,1,2] if it was rotated 4 times.
[0,1,2,4,5,6,7] if it was rotated 7 times.
Notice that rotating an array [a[0], a[1], a[2], ..., a[n-1]] 1 time results in the array [a[n-1], a[0], a[1], a[2], ..., a[n-2]].

Given the sorted rotated array nums of unique elements, return the minimum element of this array.

You must write an algorithm that runs in O(log n) time.

```python
import math
def findMinimum(nums):
    left = 0
    right = len(nums)-1
    min_ele = math.inf
    while(left<right):
        mid = (right+left)//2
        min_ele = min(min_ele, nums[mid])
        if nums[mid]>nums[right]:
            left = mid+1
        else:
            right= mid-1
    return min_ele
            


nums = [4,5,6,7,0,1,2]
print(findMinimum(nums))
```

	◦ Given a sorted array with duplicates, find the first and last position of a target element.
	◦ Example: nums = [2,4,4,4,5,6], target = 4 → output [1,3].

```python
def findFirst(nums, target):
    left = 0
    right = len(nums)-1
    first = -1
    while(left<=right):
        mid = (left+right)//2
        if nums[mid]==target:
            first = mid
            right= mid-1
        elif target<nums[mid]:
            right = mid-1
        else:
            left = mid+1
    return first

def findSecond(nums, target):
    left = 0
    right = len(nums)-1
    second = -1
    while(left<=right):
        mid = (left+right)//2
        if nums[mid]==target:
            second = mid
            left= mid+1
        elif target<nums[mid]:
            right = mid-1
        else:
            left = mid+1
    return second

def searchRange(nums, target):
    return [findFirst(nums, target), findSecond(nums, target)]


# Example
nums = [2,4,4,4,5,6]
target = 4
print(searchRange(nums, target))  # Output: [1, 3]
```

```python
def moveZeroRightSideTwoPointer(nums):
    last_non_zero = 0
    for i in range(len(nums)):
        if nums[i] != 0:
            nums[last_non_zero] = nums[i]
            if last_non_zero != i:
                nums[i] = 0
            last_non_zero += 1
    return nums

nums = [0,0,4,0,0,7,8,0,9]
print(moveZeroRightSideTwoPointer(nums))  # [4,7,8,9,0,0,0,0,0]
```

```python
def flatten_dict(d, parent_key='', sep='.'):
    items = {}
    for k, v in d.items():
        new_key = f"{parent_key}{sep}{k}" if parent_key else k
        if isinstance(v, dict):
            # Recursively flatten nested dictionary
            items.update(flatten_dict(v, new_key, sep=sep))
        else:
            items[new_key] = v
    return items

# Example usage
data = {
    "a": 1,
    "b": {
        "c": 2,
        "d": 3
    }
}
{'a': 1, 'b.c': 2, 'b.d': 3}


print(flatten_dict(data))
```

```python
from collections import deque

class TreeNode:
    def __init__(self, value, left=None, right=None):
        self.value= value
        self.left= left
        self.right = right


def levelOrderTraversal(root):
    if not root:
        return []
    result = []
    queue = deque([root])
    
    
    while queue:
        level =[]
        size = len(queue)
        for _ in range(len(size)):
            node = deque.popleft(queue)
            level.append(node.value)
            if node.left:
                queue.append(node.left.value)
                
            if node.right:
                queue.append(node.right.value)
        result.append(level)
    return result
        
        
        
            
```

```python
def longestSubStringx(s):
    freq = {}
    left =0 
    maxLength = 0
    startIndex = 0
    for right in range(len(s)):
        if s[right] in freq and freq[s[right]]>=left:
            left = freq[s[right]]+1
        freq[s[right]] = right
        currLength = right-left+1
        if maxLength<currLength:
            maxLength=currLength
            startIndex = left

    return maxLength, s[startIndex:startIndex+maxLength]
            
s =  "abcdefabcdv"
maxLength, maxString = longestSubStringx(s)
print(maxLength, maxString )
```
