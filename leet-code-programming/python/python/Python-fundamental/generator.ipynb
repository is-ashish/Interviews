"A generator is a lazy iterator — it uses yield to pause execution and save its entire frame state, producing one value at a time without loading everything into memory. Internally Python saves the frame object — local variables, instruction pointer, and stack — and resumes exactly from where it paused on each next() call."

```python
def my_generator():
    n = 1
    while n<10:
        yield n
        n +=1
    
gen = my_generator()

for i in gen:
    print("++++++++++++++++++++++++=",i)
    print(next(gen))
```

```python
def read_large_file(file_name):
    with open(file_name, "r") as f:
        for line in f:
            yield line.strip() # yield one line at a time

#usage
for line in read_large_file("big_log.txt"):
    print(line)
    
```

```python
class CountIterator:
    def __init__(self, start, end):
        self.current = start
        self.end = end
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.current <= self.end:
            val = self.current
            self.current += 1
            return val
        else:
            raise StopIteration

# Usage
for num in CountIterator(1, 10):
    print(num)
```
