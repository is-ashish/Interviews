```python
def admin_user(func):
    def wrapper(user, *args, **kwargs):
        if not user.get("Admin"):
            raise PermissionError("Admin User required to delete")
        func(user, *args, **kwargs)
    return wrapper
            
@admin_user
def deleteUser(user, user_id):
    print(f"User {user_id} is deleted by {user["Name"]}")
    
AdminUser = {"Name":"Ashish", "Admin": False}
NormalUser = {"Name":"Raj", "Admin": False}

deleteUser(AdminUser, 345)
```

```python
def require_role(allowed_roles):
    def decorator(func):
        def wrapper(UserInfo, *args, **kwargs):
            if UserInfo.get("Role") not in allowed_roles:
                raise PermissionError("Sorry, Your are not allowed to view the reports")
            func(UserInfo, *args, **kwargs)
        return wrapper
    return decorator
        
@require_role(["Manager", "Admin", "Supervisor"])
def view_report(UserInfo):
    print(f"Yeah, {UserInfo["Name"]} Can See the report")
    
AdminUser = {"Name":"Ashish", "Role": "Admin" }
ManagerUser = {"Name":"Raj", "Role":"Manager"}
NonAdminUser = {"Name":"Pankaj", "Role": "User"}


view_report(AdminUser)
```

```python
def my_logger(func):
    # The inner function accepts all arguments sent to the original function
    def wrapper(*args, **kwargs):
        print(f"Arguments passed to {func.__name__}: args={args}, kwargs={kwargs}")
        
        # Unpacking args and kwargs to execute the original function safely
        func(*args, **kwargs)
        
        # return result
    return wrapper


@my_logger
def add(a,b, task):
    print(f"{task} of numbers {a+b}")

add(3,7, task="add")

# @my_logger
# def introduce(name, role="Developer"):
#     print(f"Hi, I am {name}, a {role}.")

# introduce("Alice", role="Data Scientist")
# Prints: Arguments passed to introduce: args=('Alice',), kwargs={'role': 'Data Scientist'}


```

1. Decorators with Arguments
◦ Question: How do you write a decorator that itself takes parameters?
◦ Example: A @retry(times=3) decorator that retries a function call.
◦ Tests: Understanding closures and multiple layers of function wrapping.

```python
import random
import time

def retry(times:int):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for attempt in range(times):
                try:
                    temp = random.random() 
                    print(temp)
                    if temp < 0.8: # 80% chance of failure payment
                        raise Exception("Payment Failed")
                    else:
                        return func(*args, **kwargs)
                except Exception as e:
                    if attempt == times-1:
                        return {
                            "message":str(e),
                        }
                    time.sleep(5)
        return wrapper
    return decorator
    
@retry(times=3)
def checkPayment():
    print("Payment Success")

checkPayment()
```

Preserving Function Metadata
Question: Why does a decorated function lose its __name__ and __doc__? How do you fix it?
Answer: Use functools.wraps to preserve metadata.
Interviewers want to see if you know about wraps.

```python
import functools
def decorator(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@decorator
def my_function():
    """This is my docstring"""
    pass

print(my_function.__name__)  # wrapper
print(my_function.__doc__)   # None
```

```python
class CallCounter:
    def __init__(self, func):
        self.func = func
        self.count = 0
    def __call__(self, *args, **kwargs):
        self.count +=1
        print(f"Function Name -: {self.func.__name__}, call count-: {self.count}")
        return self.func(*args, **kwargs)
    def reset(self):
        self.count =0
    def getCounter(self):
        return self.count
    def stats(self):
        """Return current call statistics"""
        return {
            "function": self.func.__name__,
            "calls": self.count
        }


@CallCounter
def greet(name):
    pass

greet("ashish")
greet("manish")
greet("ankit")
print(greet.stats())
greet.reset()
print(greet.getCounter())
```
